```python
Part B — Data Cleaning
7. Identify all columns containing missing values.
8. Calculate the percentage of missing values in each column.
9. Handle missing values in the numerical columns appropriately.
10. Identify and remove duplicate records.
11. Check the Age column for invalid values.
12. Correct or handle patients whose age is outside a reasonable range.
13. Identify inconsistent values in categorical columns such as Gender and Department.
14. Standardize these categorical values.
```

```python

print("7. Identify all columns containing missing values.")

missing_columns = df.isnull().sum()

print(missing_columns[missing_columns > 0])
```

```python
# 8. Percentage of missing values in each column

print("===== PERCENTAGE OF MISSING VALUES =====")

missing_percentage = (df.isnull().sum() / len(df)) * 100

print(missing_percentage)
```

```python
#9. Handle missing values in the numerical columns appropriately.
import numpy as np
print("9. Handle missing values in the numerical columns appropriately.")
numeric_columns = df.select_dtypes(include=np.number).columns

print(numeric_columns)

for column in numeric_columns:
    df[column]=df[column].fillna(df[column].median())


print(df[numeric_columns].isnull().sum()) # validating the all NaNs are replaced with median
```

```python
#10. Identify and remove duplicate records.
duplicate= df.duplicated().sum()
print(duplicate)
df = df.drop_duplicates()
print(df.duplicated().sum())
```

```python
#11. Check the Age column for invalid values.
print(df[df["Age"]<0])

print("age extremely large...")
print(df[df["Age"]<100])
```

```python
#12. Correct or handle patients whose age is outside a reasonable range.
invalid_age = df[(df["Age"] < 0) | (df["Age"] > 120)]

print(invalid_age)
median_age = df.loc[
    (df["Age"] >= 0) & (df["Age"] <= 120),
    "Age"
].median()
print("median_age is" median_age)
df.loc[(df["Age"] < 0) | (df["Age"] > 120), "Age"] = median_age
print(df["Age"].describe())
```

```python
#13. Identify inconsistent values in categorical columns such as Gender and Department.
cat_col = df.select_dtypes(include="object").columns
print(cat_col)
#Reslt 
#Index(['Patient_ID', 'Gender', 'Department', 'Blood_Pressure', 'Diagnosis', 'Admission_Type', 'Readmission'],dtype='object')

for col in cat_col:
    print(df[col].value_counts(dropna=False))
```

```python
#14. Standardize these categorical values.
gender_mapping = {
    "Male": "Male",
    "male": "Male",
    "MALE": "Male",
    "FEMALE": "Female",
    "Female": "Female"
}

df["Gender"] = df["Gender"].str.strip().replace(gender_mapping)
print(df["Gender"].value_counts())
```

```python
Part C — Data Transformation
15. Create a new column called Age_Group:
< 18 Child
18–40 Young Adult
41–60 Middle Age
> 60 Senior
16. Separate Blood_Pressure into two columns:
• Systolic
• Diastolic
17. Create a High_Risk column based on suitable clinical conditions.
18. Create a Cost_Per_Day column using treatment cost and length of stay.
```

```python
#15. Create a new column called Age_Group:
bins = [-np.inf, 17, 40, 60, np.inf]

labels = [
    "Child",
    "Young Adult",
    "Middle Age",
    "Senior"
]

df["Age_Group"] = pd.cut(
    df["Age"],
    bins=bins,
    labels=labels
)
print(df[["Age", "Age_Group"]].head(20))
```

```python
16. Separate Blood_Pressure into two columns:
• Systolic
• Diastolic
```

```python
print(df["Blood_Pressure"].head())
df[["Systolic", "Diastolic"]] = (
    df["Blood_Pressure"]
    .str.split("/", expand=True)
)
print(df["Systolic"])
df["Systolic"] = pd.to_numeric(df["Systolic"], errors="coerce")
df["Diastolic"] = pd.to_numeric(df["Diastolic"], errors="coerce")
print(
    df[
        ["Blood_Pressure", "Systolic", "Diastolic"]
    ].head(10)
)
```

```python
#17. Create a High_Risk column based on suitable clinical conditions.
df["High_Risk"] = (
    (df["Systolic"] >= 140) |
    (df["Diastolic"] >= 90) |
    (df["Glucose"] >= 126) |
    (df["Cholesterol"] >= 240) |
    (df["Heart_Rate"] >= 100)
)
df["High_Risk"] = df["High_Risk"].map({
    True: "Yes",
    False: "No"
})
print(df["High_Risk"].value_counts())
```

```python
#18. Create a Cost_Per_Day column using treatment cost and length of stay.
df["Cost_Per_Day"] = np.where(
    df["Length_of_Stay"] > 0,
    df["Treatment_Cost"] / df["Length_of_Stay"],
    np.nan
)
print(
    df[
        ["Treatment_Cost", "Length_of_Stay", "Cost_Per_Day"]
    ].head(10)
)
```

```python
Part D — Analysis
19. Calculate the average age of patients.
20. Find the number of patients in each department.
21. Calculate the average reatment cost for each department.
22. Find the average length of stay for each diagnosis.
23. Identify the top 10 patients with the highest treatment cost.
24. Find the department with the highest average treatment cost.
25. Calculate the percentage of patients who were readmitted.
26. Compare readmission rates across departments.
27. Find the average cholesterol and glucose level by diagnosis.
28. Identify patients with:
• high glucose
• high cholesterol
• high heart rate
```

```python
#19. Calculate the average age of patients.
print(df["Age"].mean())
```

```python
#20. Find the number of patients in each department.
print(df["Department"].value_counts())
```

```python
avg_cost_department = (
    df.groupby("Department")["Treatment_Cost"]
      .mean()
      .sort_values(ascending=False)
)

print(avg_cost_department)
```

```python
avg_los_diagnosis = (
    df.groupby("Diagnosis")["Length_of_Stay"]
      .mean()
      .sort_values(ascending=False)
)

print(avg_los_diagnosis.round(2))
```

```python
top_10_patients = (
    df.sort_values(
        by="Treatment_Cost",
        ascending=False
    )
    .head(10)
)
print(
    top_10_patients[
        ["Patient_ID", "Treatment_Cost"]
    ]
)
```

```python
#25. Percentage of patients who were readmitted
print(df["Readmission"].value_counts(dropna=False))
readmission_percentage = (
    df["Readmission"]
    .eq("Yes")
    .mean()
    * 100
)

print(
    "Readmission Percentage:",
    round(readmission_percentage, 2),
    "%"
)
```

```python
#26. Compare readmission rates across departments
readmission_by_department = (
    df.groupby("Department")["Readmission"]
      .apply(lambda x: (x == "Yes").mean() * 100)
      .sort_values(ascending=False)
)

print(readmission_by_department.round(2))
```

```python
#27. Average cholesterol and glucose level by diagnosis
avg_clinical_by_diagnosis = (
    df.groupby("Diagnosis")
      [["Cholesterol", "Glucose"]]
      .mean()
)

print(avg_clinical_by_diagnosis.round(2))
```

```python
#28. Identify patients with high glucose, cholesterol and heart rate
# 28. High glucose
print("\nPatients with High Glucose:")
print(
    df[df["Glucose"] >= 126]
    [["Patient_ID", "Glucose"]]
)

# High cholesterol
print("\nPatients with High Cholesterol:")
print(
    df[df["Cholesterol"] >= 240]
    [["Patient_ID", "Cholesterol"]]
)

# High heart rate
print("\nPatients with High Heart Rate:")
print(
    df[df["Heart_Rate"] >= 100]
    [["Patient_ID", "Heart_Rate"]]
)
```
