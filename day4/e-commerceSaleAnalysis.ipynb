```python
Part B — Data Cleaning
7. Handle missing ratings
```

```python
print(df["Rating"].isnull().sum())
median = df["Rating"].median()
df["Rating"] = df["Rating"].fillna(median)
print(df["Rating"].isnull().sum())
```

```python
#8. Standardize city names
print(df["City"].unique())
df["City"] = (
    df["City"]
    .astype(str)
    .str.strip()
    .str.title()
)
print(df["City"].unique())
```

```python
#9. Identify invalid quantities
invalid_quantity = df[df["Quantity"] <= 0]

print(invalid_quantity)
print("Invalid quantities:", (df["Quantity"] <= 0).sum())
```

```python
df = df.drop_duplicates(subset="Order_ID", keep="first")
print(df)
```

```python
#11. Convert Order_Date to datetime

df["Order_Date"] = pd.to_datetime(
    df["Order_Date"],
    errors="coerce"
)
print(df["Order_Date"].dtype)
```

```python
Part C — Feature Engineering

Feature engineering means creating new useful columns from existing data.
12. Create Gross_Amount
```

```python
df["Gross_Amount"] = (
    df["Quantity"] *
    df["Unit_Price"]
)
print(
    df[
        ["Quantity", "Unit_Price", "Gross_Amount"]
    ].head()
)
```

```python
#3. Create Discount_Amount
df["Discount_Amount"] = (
    df["Gross_Amount"] * df["Discount"]/100
)
print(
    df[
        ["Discount", "Discount_Amount", "Gross_Amount"]
    ].head()
)
```

```python
#14. Create Net_Amount
df["Net_Amount"] = (
    df["Gross_Amount"] -
    df["Discount_Amount"]
)
print(
    df[
        ["Net_Amount", "Discount_Amount", "Gross_Amount"]
    ].head()
)
```

```python
#15. Create Rating_Category
bins = [0, 2, 3, 4, 5]

labels = [
    "Poor",
    "Average",
    "Good",
    "Excellent"
]

df["Rating_Category"] = pd.cut(
    df["Rating"],
    bins=bins,
    labels=labels,
    include_lowest=True
)
print(
    df[
        ["Rating", "Rating_Category"]
    ].head(20)
)
```

```python
#16. Extract Year, Month, Day and Day of Week
df["Year"] = df["Order_Date"].dt.year
df["Month"] = df["Order_Date"].dt.month
df["Day"] = df["Order_Date"].dt.day
df["Day_of_Week"] = df["Order_Date"].dt.day_name()
print(
    df[
        [
            "Order_Date",
            "Year",
            "Month",
            "Day",
            "Day_of_Week"
        ]
    ].head(10)
)
```

```python
Part D — Business Analysis

Now we use the cleaned and transformed data to answer business questions.
    17. Calculate total revenue
```

```python
total_revenue = df["Net_Amount"].sum()

print("Total Revenue:", total_revenue)
```

```python
#18. Calculate total number of orders
order = df["Order_ID"].nunique()
print(order)
```

```python
#19. Calculate Average Order Value
average_order_value = (
    df["Net_Amount"].sum() /
    df["Order_ID"].nunique()
)

print(
    "Average Order Value:",
    round(average_order_value, 2)
)
```

```python
# 20. Revenue by Category
revenue_by_category = (
    df.groupby("Product_Category")
      ["Net_Amount"]
      .sum()
      .sort_values(ascending=False)
)

print("\nRevenue by Category:")
print(revenue_by_category.round(2))
```

```python
# 21. Top 10 Products
top_products = (
    df.groupby("Product")["Net_Amount"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

print("\nTop 10 Products:")
print(top_products.round(2))
```

```python
# 22. Highest Revenue City
city_revenue = (
    df.groupby("City")["Net_Amount"]
      .sum()
      .sort_values(ascending=False)
)

print("\nRevenue by City:")
print(city_revenue.round(2))

print("\nHighest Revenue City:")
print(city_revenue.idxmax())
```

```python
# 23. Most Used Payment Method
payment_method = (
    df["Payment_Method"]
    .value_counts()
)

print("\nPayment Methods:")
print(payment_method)

print("\nMost Used Payment Method:")
print(payment_method.idxmax())
```

```python




# 24. Average Rating by Category
average_rating = (
    df.groupby("Product_Category")
      ["Rating"]
      .mean()
)

print("\nAverage Rating by Category:")
print(average_rating.round(2))
```

```python


# 25. Monthly Revenue
monthly_revenue = (
    df.groupby(
        df["Order_Date"].dt.to_period("M")
    )["Net_Amount"]
    .sum()
)

print("\nMonthly Revenue:")
print(monthly_revenue.round(2))
```

```python


# 26. Highest Revenue Month
print("\nHighest Revenue Month:")
print(monthly_revenue.idxmax())

print("Revenue:")
print(round(monthly_revenue.max(), 2))
```

```python


# 28. City Comparison
print("\nCity Revenue:")
print(city_revenue.round(2))

city_percentage = (
    city_revenue /
    city_revenue.sum() *
    100
)

print("\nCity Revenue Percentage:")
print(city_percentage.round(2))

# 29. Highest AOV Category
category_aov = (
    df.groupby("Product_Category")
      ["Net_Amount"]
      .mean()
      .sort_values(ascending=False)
)

print("\nAverage Order Value by Category:")
print(category_aov.round(2))

print("\nHighest AOV Category:")
print(category_aov.idxmax())

print("AOV:")
print(round(category_aov.max(), 2))
```
